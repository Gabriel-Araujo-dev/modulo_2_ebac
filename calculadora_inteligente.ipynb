#Calculadora


```python
# ==========================================
#           CALCULADORA EM PYTHON
# ==========================================

print("===== CALCULADORA =====")

# O programa continuará executando até que o usuário escolha sair.
while True:

    # Exibe o menu de operações disponíveis.
    print("\nEscolha uma operação:")
    print("1 - Adição")
    print("2 - Subtração")
    print("3 - Multiplicação")
    print("4 - Divisão")
    print("5 - Sair")

    # Recebe a opção escolhida pelo usuário.
    opcao = input("Digite a opção desejada: ")

    # Encerra o programa caso o usuário escolha a opção 5.
    if opcao == "5":
        print("\nCalculadora encerrada. Até mais!")
        break

    # Verifica se a opção informada é válida.
    if opcao not in ["1", "2", "3", "4"]:
        print("\nOpção inválida! Tente novamente.")
        continue

    # Recebe os dois números e converte os valores para float.
    numero1 = float(input("Digite o primeiro número: "))
    numero2 = float(input("Digite o segundo número: "))

    # Realiza a operação escolhida pelo usuário.
    if opcao == "1":
        resultado = numero1 + numero2
        simbolo = "+"

    elif opcao == "2":
        resultado = numero1 - numero2
        simbolo = "-"

    elif opcao == "3":
        resultado = numero1 * numero2
        simbolo = "×"

    elif opcao == "4":

        # Evita uma divisão por zero.
        if numero2 == 0:
            print("\nErro: não é possível dividir por zero.")
            continue

        resultado = numero1 / numero2
        simbolo = "÷"

    # Exibe a operação e o resultado para o usuário.
    print(f"\nResultado: {numero1} {simbolo} {numero2} = {resultado}")
```
